```python
"""Top Product Analysis
Business Question:
Management wants to identify the products generating the highest Net Revenue and Gross Profit.

For each product, calculate:
- ProductName
- Category
- TotalQuantity
- NetRevenue
- GrossProfit
- ProfitMargin"""
```

```python
ProductAnalysis= Sales.groupby(["ProductName","Category"]).agg(
    TotalQuantity=("Quantity","sum"),
    NetRevenue=("NetRevenue","sum"),
    GrossProfit=("GrossProfit","sum"),
)
ProductAnalysis["ProfitMargin"]= ProductAnalysis["GrossProfit"]/ProductAnalysis["NetRevenue"]*100
ProductAnalysis.sort_values("NetRevenue",ascending=False).head(5)
```

```python
#Top 5 products by GrossProfit.
ProductAnalysis["GrossProfit"].sort_values(ascending= False).head(5)
```

```python
#High Revenue vs Low Margin(low margin as below 25%.)
ProductAnalysis[ProductAnalysis["ProfitMargin"] < 25]
```
